```python
import pandas as pd

# Load the three datasets
by_df = pd.read_csv("BY_DATASET.csv")
cd_df = pd.read_csv("CD_DATASET.csv")
cdby_df = pd.read_csv("CD_BY_DATASET.csv")

# Drop percentage helper columns from CD+BY dataset (not needed)
cdby_df.drop(columns=["CD_%", "BY_%"], inplace=True)

# Encode indicator columns using only BY and CD
# BY_DATASET: BY=1, CD=0
by_df["BY"] = by_df["BY"].apply(lambda x: 1 if str(x).strip().lower() == "yes" else 0)
by_df["CD"] = 0

# CD_BY_DATASET: if CD+BY is Yes, then both BY=1 and CD=1
cdby_flag = cdby_df["CD + BY"].apply(lambda x: 1 if str(x).strip().lower() == "yes" else 0)
cdby_df["BY"] = cdby_flag
cdby_df["CD"] = cdby_flag
cdby_df.drop(columns=["CD + BY"], inplace=True)

# CD_DATASET: CD=1, BY=0
cd_df["CD"] = cd_df["CD"].apply(lambda x: 1 if str(x).strip().lower() == "yes" else 0)
cd_df["BY"] = 0

# Define a consistent column order (no CD+BY column)
col_order = ["main_name", "BY", "CD", "SW_%", "KW_%", "LF_%", "ST_%", "Time", "Volume"]

merged = pd.concat(
    [by_df[col_order], cdby_df[col_order], cd_df[col_order]],
    ignore_index=True,
)

print(merged.shape)
merged.head(10)
```

```python
merged.to_csv("MERGED_DATASET.csv", index=False)
```

```python
import numpy as np

# Material properties from Thesis Tables 4.1 and 4.2
# TS / VS are fractions of wet mass; C / N are fractions of dry mass (TS)
MATERIAL_PROPERTIES = {
    'SW': {'type': 'substrate', 'TS': 0.0767, 'VS': 0.0650, 'C': 0.3079, 'N': 0.0769},
    'KW': {'type': 'substrate', 'TS': 0.0957, 'VS': 0.0840, 'C': 0.4800, 'N': 0.0260},
    'ST': {'type': 'substrate', 'TS': 0.9077, 'VS': 0.7797, 'C': 0.3745, 'N': 0.00412},
    'LF': {'type': 'substrate', 'TS': 0.9073, 'VS': 0.7377, 'C': 0.4810, 'N': 0.0120},
    'CD': {'type': 'inoculum', 'TS': 0.2260, 'VS': 0.1777, 'C': 0.4048, 'N': 0.0137},
    'BY': {'type': 'inoculum', 'TS': 0.7500, 'VS': 0.1800, 'C': 0.4830, 'N': 0.1060},
}


def calculate_wet_masses(substrate_fractions, inoculum_types, total_mass=200.0, target_isr=2.0):
    """Back-calculates wet masses (g) for each component in a 200 g reactor at ISR=2:1 VS."""
    vs_sub = sum(frac * MATERIAL_PROPERTIES[mat]['VS'] for mat, frac in substrate_fractions.items())
    n_inoc = len(inoculum_types)
    vs_inoc = sum((1.0 / n_inoc) * MATERIAL_PROPERTIES[mat]['VS'] for mat in inoculum_types)

    k_ratio = target_isr * vs_sub / vs_inoc
    mass_sub = total_mass / (1 + k_ratio)
    mass_inoc = total_mass - mass_sub

    wet_masses = {mat: mass_sub * frac for mat, frac in substrate_fractions.items()}
    for mat in inoculum_types:
        wet_masses[mat] = mass_inoc / n_inoc
    return wet_masses


def engineer_features(wet_masses):
    """Returns TS%, VS%, C/N ratio, and ISR for a given wet-mass composition."""
    total_wet = sum(wet_masses.values())
    total_ts = total_vs = total_c = total_n = 0.0
    vs_inoc = vs_sub = 0.0

    for mat, mass in wet_masses.items():
        props = MATERIAL_PROPERTIES[mat]
        ts_mass = mass * props['TS']
        vs_mass = mass * props['VS']

        total_ts += ts_mass
        total_vs += vs_mass
        total_c += ts_mass * props['C']
        total_n += ts_mass * props['N']

        if props['type'] == 'inoculum':
            vs_inoc += vs_mass
        else:
            vs_sub += vs_mass

    return {
        'TS_mix_pct': (total_ts / total_wet) * 100,
        'VS_mix_pct': (total_vs / total_wet) * 100,
        'CN_ratio_mix': total_c / total_n if total_n > 0 else 0.0,
        'ISR': vs_inoc / vs_sub if vs_sub > 0 else 0.0,
    }


def row_features(row):
    """Derive feature dict from a single merged-dataset row."""
    # Substrate fractions (% -> fraction)
    substrate_fractions = {}
    for col, mat in [('SW_%', 'SW'), ('KW_%', 'KW'), ('LF_%', 'LF'), ('ST_%', 'ST')]:
        value = row[col]
        if pd.notna(value) and value > 0:
            substrate_fractions[mat] = value / 100.0

    # Inoculum mapping without CD+BY column:
    # BY=1 and CD=1 means both are present.
    inoculum_types = []
    if row['BY'] == 1:
        inoculum_types.append('BY')
    if row['CD'] == 1:
        inoculum_types.append('CD')

    if not substrate_fractions or not inoculum_types:
        return {
            'TS_mix_pct': np.nan,
            'VS_mix_pct': np.nan,
            'CN_ratio_mix': np.nan,
            'ISR': np.nan,
        }

    wet_masses = calculate_wet_masses(substrate_fractions, inoculum_types)
    return engineer_features(wet_masses)


feature_cols = merged.apply(row_features, axis=1, result_type='expand')
merged_fe = pd.concat([merged, feature_cols], axis=1)

print(f"New features added: {list(feature_cols.columns)}")
print(f"Dataset shape: {merged_fe.shape}")
merged_fe.head(10)
```

```python
# Cell 12: remove CD+BY from existing merged_fe only
# Assumes merged_fe already exists from Cell 10
if "CD + BY" in merged_fe.columns:
    # Ensure BY and CD columns exist
    if "BY" not in merged_fe.columns:
        merged_fe["BY"] = 0
    if "CD" not in merged_fe.columns:
        merged_fe["CD"] = 0

    cdby_mask = merged_fe["CD + BY"].fillna(0).eq(1)
    merged_fe.loc[cdby_mask, "BY"] = 1
    merged_fe.loc[cdby_mask, "CD"] = 1
    merged_fe.drop(columns=["CD + BY"], inplace=True)
    print("CD + BY removed from merged_fe; mapped to BY=1 and CD=1.")
else:
    print("No CD + BY column found in merged_fe (nothing to remove).")

print("merged_fe columns:", list(merged_fe.columns))
merged_fe.head()
```
