```python
import pandas as pd
import re
import os

files_names = [
    'CD + 40% SW + 60% KW',
    'CD + 60% SW + 30% KW + 10% LF',
    'CD + 60% SW + 30% KW + 10% ST',
    'CD + 62.5% SW + 30% KW + 7.5% LF',
    'CD + 65% SW + 30% KW + 5% LF',
    'CD + 65% SW + 30% KW + 5% ST',
    'CD + 68.5% SW + 30% KW + 1.5% LF',
    'CD + 68.5% SW + 30% KW + 1.5% ST',
    'CD + 70% SW + 30% KW',
    'CD + KW',
    'CD + SW'
]

def parse_filename(filename):
    """
    Parses the filename to extract categorical presence and numerical percentages.
    """
    features = {'main_name': filename, 'CD': 'No'}
    
    # Split the string by the '+' delimiter
    parts = [p.strip() for p in filename.split('+')]
    
    for part in parts:
        if part == 'CD':
            features['CD'] = 'Yes'
        else:
            # Regex to match explicit percentages (e.g., '62.5% SW')
            pct_match = re.match(r'(\d+\.?\d*)%\s*([A-Za-z]+)', part)
            if pct_match:
                val, mat = pct_match.groups()
                features[f'{mat}_%'] = float(val)
            else:
                # Match implied 100% additive components (e.g., 'KW' in 'CD + KW')
                mat_match = re.match(r'([A-Za-z]+)', part)
                if mat_match:
                    mat = mat_match.group(1)
                    features[f'{mat}_%'] = 100.0 
                    
    return features

dataframes = []

for name in files_names:
    csv_path = f"{name}.csv"
    
    # In a real run, you'd want to handle FileNotFoundError appropriately.
    if os.path.exists(csv_path):
        df = pd.read_csv(csv_path)
        
        parsed_data = parse_filename(name)
        
        # Broadcast the parsed scalar values to the entire DataFrame
        for col_name, value in parsed_data.items():
            df[col_name] = value
            
        dataframes.append(df)
    else:
        print(f"Skipping {csv_path} - File not found.")

# Concatenate into a single master dataset if files were found
if dataframes:
    master_df = pd.concat(dataframes, ignore_index=True)
    
    # Any material not present in a specific file will yield NaN. 
    # Impute these with 0.0 for the percentage columns.
    percentage_cols = [c for c in master_df.columns if c.endswith('_%')]
    master_df[percentage_cols] = master_df[percentage_cols].fillna(0.0)
    
    # Display the unique generated features to verify extraction
    print(master_df.drop_duplicates(subset=['main_name'])[
        ['main_name', 'CD'] + percentage_cols
    ])
    
    # master_df.to_csv('combined_extracted_data.csv', index=False)
```

```python
import pandas as pd

# Assuming the concatenated dataframe from the previous step is named 'master_df'
# and contains 'Time', 'Volume', 'main_name', plus the metadata columns.

# 1. Filter out rows with negative Time or Volume. 
# Using (Time >= 0) & (Volume >= 0) removes a row if *either* is negative. 
# If you strictly meant omitting only when *both* are negative concurrently, 
# you would use: master_df[~((master_df['Time'] < 0) & (master_df['Volume'] < 0))]
master_df = master_df[(master_df['Time'] >= 0) & (master_df['Volume'] >= 0)].copy()

# 2. Isolate the metadata columns (everything except Time and Volume)
meta_cols = [col for col in master_df.columns if col not in ['Time', 'Volume']]

# 3. Extract the unique metadata mapping for each 'main_name'
# This ensures the new 0,0 rows inherit the correct CD status and percentages
zero_rows = master_df[meta_cols].drop_duplicates(subset=['main_name']).copy()

# 4. Assign Time = 0 and Volume = 0
zero_rows['Time'] = 0.0
zero_rows['Volume'] = 0.0

# 5. Concatenate the zero-initialized rows with the filtered dataset
master_df = pd.concat([zero_rows, master_df], ignore_index=True)

# 6. Drop potential duplicates (in case a Time=0, Volume=0 row already existed)
# and sort the dataframe to maintain chronological order per file
master_df = master_df.drop_duplicates().sort_values(by=['main_name', 'Time']).reset_index(drop=True)

# Verify the insertion
print(master_df.groupby('main_name').head(2))
```
