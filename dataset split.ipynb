```python
import pandas as pd

# Define the split percentage (60% Train, 40% Test based on max Time)
SPLIT_RATIO = 0.60

train_list = []
test_list = []
data=pd.read_csv("MERGED FINAL.csv")
print("Splitting data by maximum time per condition...")

for name, group in data.groupby('main_name'):
    # Ensure data is sorted by time chronologically just in case
    group = group.sort_values(by='Time')
    
    # Calculate the maximum time observed for this reactor
    max_time = group['Time'].max()
    
    # Define the chronological cutoff point
    cutoff_time = max_time * SPLIT_RATIO
    
    # Split the temporal data exactly by the calculated time boundary
    train_group = group[group['Time'] <= cutoff_time]
    test_group = group[group['Time'] > cutoff_time]
    
    train_list.append(train_group)
    test_list.append(test_group)

# Combine the individual parts back into comprehensive DataFrames
train_df = pd.concat(train_list, ignore_index=True)
test_df = pd.concat(test_list, ignore_index=True)

# Save the resulting files as CSVs
train_file = f"train_split_temporal_{int(SPLIT_RATIO*100)} no main name.csv"
test_file = f"test_split_temporal_{int((1-SPLIT_RATIO)*100)} no main name.csv"
test_df.drop(columns=['main_name'], inplace=True)
train_df.drop(columns=['main_name'], inplace=True)
train_df.to_csv(train_file, index=False)
test_df.to_csv(test_file, index=False)

print(f"Temporal Data Split Completed Successfully!")
print(f"Saved Train set ({len(train_df)} rows): {train_file}")
print(f"Saved Test set ({len(test_df)} rows): {test_file}")
```
