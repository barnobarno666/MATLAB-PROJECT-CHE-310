```python
import pandas as pd
import numpy as np

# Load your dataset

# 1. Define the Elemental Compositions from Thesis Table 4.2
# Format: {'Material': {'C': %C, 'H': %H, 'O': %O, 'N': %N, 'S': %S}}
elements = {
    'SW': {'C': 30.79, 'H': 5.37, 'O': 39.00, 'N': 7.69,  'S': 9.100},
    'KW': {'C': 48.00, 'H': 6.40, 'O': 37.60, 'N': 2.60,  'S': 0.400},
    'ST': {'C': 37.45, 'H': 5.10, 'O': 33.81, 'N': 0.412, 'S': 0.061},
    'LF': {'C': 48.10, 'H': 6.10, 'O': 44.50, 'N': 1.20,  'S': 0.100},
    'CD': {'C': 40.48, 'H': 5.29, 'O': 29.66, 'N': 1.37,  'S': 0.211},
    'BY': {'C': 48.30, 'H': 6.00, 'O': 34.50, 'N': 10.60, 'S': 0.200}
}

def calculate_mixture_elements(row):
    # Substrate fractions (convert percentages to 0-1 decimals)
    f_sw = row['SW_%'] / 100.0
    f_kw = row['KW_%'] / 100.0
    f_lf = row['LF_%'] / 100.0
    f_st = row['ST_%'] / 100.0
    
    # Inoculum indicators (0 or 1)
    has_by = row['BY']
    has_cd = row['CD']
    
    # Determine Inoculum Fractions
    # If both BY and CD are present, they split the inoculum portion (50/50)
    total_inoculum_types = has_by + has_cd
    f_by = (1.0 / total_inoculum_types) if (has_by and total_inoculum_types > 0) else 0.0
    f_cd = (1.0 / total_inoculum_types) if (has_cd and total_inoculum_types > 0) else 0.0
    
    # Weighting based on ISR (Inoculum to Substrate Ratio)
    # Total mass parts = ISR + 1 (e.g., if ISR is 2.0, Inoculum=2 parts, Substrate=1 part)
    isr = row['ISR']
    w_inoculum = isr / (isr + 1.0)
    w_substrate = 1.0 / (isr + 1.0)
    
    # Calculate final weighted percentages for each element
    mix_props = {}
    for el in ['C', 'H', 'O', 'N', 'S']:
        # Substrate elemental contribution
        sub_elem = (f_sw * elements['SW'][el] + 
                    f_kw * elements['KW'][el] + 
                    f_lf * elements['LF'][el] + 
                    f_st * elements['ST'][el])
        
        # Inoculum elemental contribution
        ino_elem = (f_by * elements['BY'][el] + 
                    f_cd * elements['CD'][el])
        
        # Final weighted mixture percentage
        mix_props[f"{el}_mix_pct"] = (sub_elem * w_substrate) + (ino_elem * w_inoculum)
        
    return pd.Series(mix_props)

# 2. Apply the calculation to generate the new columns
elemental_cols = df.apply(calculate_mixture_elements, axis=1)

# 3. Concatenate the new elemental features to your original dataset
df_enriched = pd.concat([df, elemental_cols], axis=1)

# Check the new elemental 
data=df_enriched.copy()
data.head()
```

```python
import numpy as np
from scipy.optimize import curve_fit
import matplotlib.pyplot as plt
from sklearn.metrics import r2_score
from IPython.display import display
import pandas as pd

def gompertz(t, P, R_m, lam):
    """
    Modified Gompertz equation for biogas production.
    """
    e = np.exp(1)
    return P * np.exp(-np.exp((R_m * e / P) * (lam - t) + 1))

features = []  
targets = []   
reactor_names = []
fit_metrics = []

for name, group in data.groupby('main_name'):
    t_raw = group['Time'].values
    V_raw = group['Volume'].values
    
    # helper function to handle duplicate columns if cell was run multiple times
    def get_scalar(col):
        val = group[col].iloc[0]
        if isinstance(val, pd.Series):
            return float(val.iloc[0])
        return float(val)
        
    ts_mix = get_scalar('TS_mix_pct')
    vs_mix = get_scalar('VS_mix_pct')
    cn_mix = get_scalar('CN_ratio_mix')
    isr = get_scalar('ISR')
    
    sw = get_scalar('SW_%')
    by = get_scalar('BY')
    cd = get_scalar('CD')
    kw = get_scalar('KW_%')
    lf = get_scalar('LF_%')
    st = get_scalar('ST_%')
    h_mix = get_scalar('H_mix_pct')
    c_mix = get_scalar('C_mix_pct')
    o_mix = get_scalar('O_mix_pct')
    n_mix = get_scalar('N_mix_pct')
    
    P_guess = np.max(V_raw) if np.max(V_raw) > 0 else 1.0
    
    dt = np.diff(t_raw)
    dV = np.diff(V_raw)
    dt[dt == 0] = 1e-6
    slopes = dV / dt
    Rm_guess = np.max(slopes) if np.max(slopes) > 0 else 1.0
    
    threshold = 0.05 * P_guess
    lam_indices = np.where(V_raw > threshold)[0]
    lam_guess = t_raw[lam_indices[0]] if len(lam_indices) > 0 else 0.0
    
    try:
        popt, pcov = curve_fit(
            gompertz, 
            t_raw, 
            V_raw, 
            p0=[P_guess, Rm_guess, lam_guess],
            bounds=(0, [np.inf, np.inf, np.inf]),
            maxfev=10000
        )
        P_ext, Rm_ext, lam_ext = popt
        
        V_pred = gompertz(t_raw, P_ext, Rm_ext, lam_ext)
        r2 = r2_score(V_raw, V_pred)
        
        features.append([ts_mix, vs_mix, cn_mix, isr, by, cd, h_mix, c_mix, o_mix, n_mix]) 
        targets.append([P_ext, Rm_ext, lam_ext])
        reactor_names.append(name)
        fit_metrics.append({'Composition': name, 'R2': r2})
        
    except Exception as e:
        print(f"Failed to fit Gompertz curve for {name}: {e}")

X = np.array(features, dtype=float)
Y = np.array(targets, dtype=float)

print(f"Successfully extracted parameters for {len(X)} reactors.")
print("Feature matrix shape:", X.shape)
print("Target matrix shape:", Y.shape)

metrics_df = pd.DataFrame(fit_metrics)
print("\nGompertz Curve Fit R2 Scores:")
display(metrics_df)
```

```python
from sklearn.svm import SVR
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import RBF, ConstantKernel as C, WhiteKernel
from sklearn.multioutput import MultiOutputRegressor
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_absolute_error, r2_score
import numpy as np

# -------------------------------------------------------------
# Support Vector Regressor (SVR) & Gaussian Process Regressor (GPR)
# Training Pipeline Strategy using Log1p Transforms
# -------------------------------------------------------------

# Apply log1p transform to smooth wide distributions and prevent strictly negative predictions
X_log = np.log1p(X)
Y_log = np.log1p(Y)

scaler_X = StandardScaler()
scaler_Y = StandardScaler()

X_scaled = scaler_X.fit_transform(X_log)
Y_scaled = scaler_Y.fit_transform(Y_log)

X_train, X_test, Y_train, Y_test = train_test_split(X_scaled, Y_scaled, test_size=0.2, random_state=42)

# Define Support Vector Regressor
svr_base = SVR(
    kernel='rbf',
    C=10.0,
    epsilon=0.1,
    gamma='scale'
)

# Define Gaussian Process Regressor
kernel = C(1.0, (1e-3, 1e3)) * RBF(10, (1e-2, 1e2)) + WhiteKernel(noise_level=1e-2)
gpr_base = GaussianProcessRegressor(kernel=kernel, n_restarts_optimizer=5, random_state=42)

# Use MultiOutputRegressor to predict P, R_m, and lambda simultaneously
svr_model = MultiOutputRegressor(svr_base)
gpr_model = MultiOutputRegressor(gpr_base)

print("Training Multi-Output SVR and GPR Models on X = [TS, VS, C/N, ISR] to predict Y = [P, Rm, λ]...")
svr_model.fit(X_train, Y_train)
gpr_model.fit(X_train, Y_train)

# -------------------------------------------------------------
# Evaluation Strategy
# -------------------------------------------------------------
Y_pred_train_svr = svr_model.predict(X_train)
Y_pred_test_svr = svr_model.predict(X_test)

Y_pred_train_gpr = gpr_model.predict(X_train)
Y_pred_test_gpr = gpr_model.predict(X_test)

# Unscale back and Inverse Log (expm1) back to biological units
Y_test_orig = np.expm1(scaler_Y.inverse_transform(Y_test))

# SVR
Y_pred_test_orig_svr = np.expm1(scaler_Y.inverse_transform(Y_pred_test_svr))
mae_test_svr = mean_absolute_error(Y_test_orig, Y_pred_test_orig_svr)
r2_test_svr = r2_score(Y_test_orig, Y_pred_test_orig_svr)

print(f"\nSVR Test Performance on Kinematic Parameters:")
print(f"Mean Absolute Error: {mae_test_svr:.3f}")
print(f"R-squared Score: {r2_test_svr:.3f}")

# GPR
Y_pred_test_orig_gpr = np.expm1(scaler_Y.inverse_transform(Y_pred_test_gpr))
mae_test_gpr = mean_absolute_error(Y_test_orig, Y_pred_test_orig_gpr)
r2_test_gpr = r2_score(Y_test_orig, Y_pred_test_orig_gpr)

print(f"\nGPR Test Performance on Kinematic Parameters:")
print(f"Mean Absolute Error: {mae_test_gpr:.3f}")
print(f"R-squared Score: {r2_test_gpr:.3f}")
```

```python
import numpy as np
import pandas as pd
from xgboost import XGBRegressor
from sklearn.svm import SVR
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import RBF, ConstantKernel as C, WhiteKernel
from sklearn.multioutput import MultiOutputRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import r2_score, mean_absolute_error
import matplotlib.pyplot as plt
import seaborn as sns
import math
from IPython.display import display

sns.set_theme(style="whitegrid", context="talk")

unique_names = np.array(reactor_names)

leave_out_results = []
plot_data_list = []

# Features for the "Normal XGBoost" (Direct Time -> Volume mapping)
normal_xgb_features = ['Time', 'SW_%', 'BY', 'CD', 'KW_%', 'LF_%', 'ST_%']

print("Running Leave-One-Out Cross-Validation (Full Curve Generation)...")

# Perform Leave-One-Out Cross-Validation at the Reactor Level
for dropped_name in set(unique_names):
    
    # Setup masks for HYBRID Models (Kinetics Prediction: TS, VS, CN, ISR -> P, Rm, lam)
    train_mask = unique_names != dropped_name
    test_mask = unique_names == dropped_name
    
    X_hybrid_train, Y_hybrid_train = X[train_mask], Y[train_mask]
    X_hybrid_test, Y_hybrid_test = X[test_mask], Y[test_mask]
    
    # Setup masks/data for NORMAL XGBOOST (Time Series Prediction: Composition + Time -> Volume)
    df_train_normal = data[data['main_name'] != dropped_name]
    df_test_normal = data[data['main_name'] == dropped_name].sort_values(by='Time')
    
    # ---------------------------------------------------------
    # 1. Train & Predict: HYBRID GOMPERTZ MODELS (SVR & GPR)
    # ---------------------------------------------------------
    # Apply log1p scale to handle skewness
    X_hybrid_train_log = np.log1p(X_hybrid_train)
    Y_hybrid_train_log = np.log1p(Y_hybrid_train)
    X_hybrid_test_log = np.log1p(X_hybrid_test)
    
    # Scale specific to this fold to avoid data leakage
    sc_X = StandardScaler()
    sc_Y = StandardScaler()
    
    X_hybrid_train_scaled = sc_X.fit_transform(X_hybrid_train_log)
    Y_hybrid_train_scaled = sc_Y.fit_transform(Y_hybrid_train_log)
    X_hybrid_test_scaled = sc_X.transform(X_hybrid_test_log)
    
    # --- Train SVR ---
    model_hybrid_svr = MultiOutputRegressor(SVR(kernel='rbf', C=10.0, epsilon=0.1, gamma='scale'))
    model_hybrid_svr.fit(X_hybrid_train_scaled, Y_hybrid_train_scaled)
    
    # Predict kinetic parameters for the dropped reactor (SVR)
    y_test_pred_scaled_svr = model_hybrid_svr.predict(X_hybrid_test_scaled)
    y_test_pred_svr = np.expm1(sc_Y.inverse_transform(y_test_pred_scaled_svr))
    P_pred_svr, Rm_pred_svr, lam_pred_svr = y_test_pred_svr[0]
    
    # --- Train GPR ---
    kernel = C(1.0, (1e-3, 1e3)) * RBF(10, (1e-2, 1e2)) + WhiteKernel(noise_level=1e-2)
    model_hybrid_gpr = MultiOutputRegressor(GaussianProcessRegressor(kernel=kernel, n_restarts_optimizer=5, random_state=42))
    model_hybrid_gpr.fit(X_hybrid_train_scaled, Y_hybrid_train_scaled)
    
    # Predict kinetic parameters for the dropped reactor (GPR)
    y_test_pred_scaled_gpr = model_hybrid_gpr.predict(X_hybrid_test_scaled)
    y_test_pred_gpr = np.expm1(sc_Y.inverse_transform(y_test_pred_scaled_gpr))
    P_pred_gpr, Rm_pred_gpr, lam_pred_gpr = y_test_pred_gpr[0]
    
    # Reconstruct the volume curve using Gompertz
    t_actual = df_test_normal['Time'].values
    V_actual = df_test_normal['Volume'].values
    
    V_pred_hybrid_svr = gompertz(t_actual, P_pred_svr, Rm_pred_svr, lam_pred_svr)
    V_pred_hybrid_gpr = gompertz(t_actual, P_pred_gpr, Rm_pred_gpr, lam_pred_gpr)
    
    # ---------------------------------------------------------
    # 2. Train & Predict: NORMAL XGBOOST MODEL (Baseline)
    # ---------------------------------------------------------
    model_normal = XGBRegressor(
        objective="reg:squarederror",
        n_estimators=300,
        max_depth=6,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.9,
        random_state=42
    )
    
    X_norm_train = df_train_normal[normal_xgb_features]
    y_norm_train = df_train_normal['Volume']
    
    X_norm_test = df_test_normal[normal_xgb_features]
    
    model_normal.fit(X_norm_train, y_norm_train)
    V_pred_normal = model_normal.predict(X_norm_test)
    
    # ---------------------------------------------------------
    # 3. Evaluate Metrics
    # ---------------------------------------------------------
    hybrid_svr_r2 = r2_score(V_actual, V_pred_hybrid_svr) if len(V_actual) > 1 else np.nan
    hybrid_svr_mae = mean_absolute_error(V_actual, V_pred_hybrid_svr) if len(V_actual) > 0 else np.nan
    
    hybrid_gpr_r2 = r2_score(V_actual, V_pred_hybrid_gpr) if len(V_actual) > 1 else np.nan
    hybrid_gpr_mae = mean_absolute_error(V_actual, V_pred_hybrid_gpr) if len(V_actual) > 0 else np.nan
    
    normal_r2 = r2_score(V_actual, V_pred_normal) if len(V_actual) > 1 else np.nan
    normal_mae = mean_absolute_error(V_actual, V_pred_normal) if len(V_actual) > 0 else np.nan
    
    leave_out_results.append({
        "main_name": dropped_name,
        "n_test_rows": len(V_actual),
        "Hybrid_SVR_Test_R2": hybrid_svr_r2,
        "Hybrid_SVR_Test_MAE": hybrid_svr_mae,
        "Hybrid_GPR_Test_R2": hybrid_gpr_r2,
        "Hybrid_GPR_Test_MAE": hybrid_gpr_mae,
        "Normal_Test_R2": normal_r2,
        "Normal_Test_MAE": normal_mae
    })
    
    plot_data_list.append({
        "name": dropped_name,
        "t": t_actual,
        "V_actual": V_actual,
        "V_hybrid_svr": V_pred_hybrid_svr,
        "V_hybrid_gpr": V_pred_hybrid_gpr,
        "V_normal": V_pred_normal,
        "Hybrid_SVR_R2": hybrid_svr_r2,
        "Hybrid_SVR_MAE": hybrid_svr_mae,
        "Hybrid_GPR_R2": hybrid_gpr_r2,
        "Hybrid_GPR_MAE": hybrid_gpr_mae,
        "Normal_R2": normal_r2,
        "Normal_MAE": normal_mae
    })

results_df = pd.DataFrame(leave_out_results).sort_values("Hybrid_SVR_Test_R2", ascending=False).reset_index(drop=True)
print("\n========= Leave-One-Out Unseen Evaluation: Hybrid Gompertz vs Normal XGBoost =========")
display(results_df[[
    'main_name', 'n_test_rows', 
    'Hybrid_SVR_Test_R2', 'Hybrid_GPR_Test_R2', 'Normal_Test_R2',
    'Hybrid_SVR_Test_MAE', 'Hybrid_GPR_Test_MAE', 'Normal_Test_MAE'
]].head(10))

# -------------------------------------------------------------
# 4. HUGE MULTI-SUBPLOT FOR ALL LEFT-OUT COMPOSITIONS
# -------------------------------------------------------------
num_plots = len(plot_data_list)
if num_plots > 0:
    cols = 4  # 4 columns wide
    rows = math.ceil(num_plots / cols)
    
    fig, axes = plt.subplots(rows, cols, figsize=(30, rows * 5.5))
    axes = axes.flatten()

    for idx, pdata in enumerate(plot_data_list):
        ax = axes[idx]
        
        t = pdata["t"]
        if len(t) == 0:
            ax.set_visible(False)
            continue
            
        # Plot Out-Of-Sample actual data
        ax.scatter(t, pdata["V_actual"], color='#0284c7', s=25, label='Actual Left-Out Data', alpha=0.6)
        
        # Plot model predictions
        ax.plot(t, pdata["V_hybrid_svr"], color='#0f172a', linewidth=2.5, linestyle='-', label='Hybrid Gompertz (SVR)')
        ax.plot(t, pdata["V_hybrid_gpr"], color='#8b5cf6', linewidth=2.5, linestyle='--', label='Hybrid Gompertz (GPR)')
        ax.plot(t, pdata["V_normal"], color='#16a34a', linewidth=2.5, linestyle='-.', label='Normal XGBoost')
            
        # Display Metrics in the plot title
        title_text = f"{pdata['name']}\n"
        title_text += f"SVR R²={pdata['Hybrid_SVR_R2']:.2f}, MAE={pdata['Hybrid_SVR_MAE']:.1f} | "
        title_text += f"GPR R²={pdata['Hybrid_GPR_R2']:.2f}, MAE={pdata['Hybrid_GPR_MAE']:.1f}\n"
        title_text += f"Norm R²={pdata['Normal_R2']:.2f}, MAE={pdata['Normal_MAE']:.1f}"
        
        ax.set_title(title_text, fontsize=10, weight='bold')
        ax.set_xlabel("Time", fontsize=9)
        ax.set_ylabel("Volume", fontsize=9)
        ax.tick_params(axis='both', which='major', labelsize=8)
        
        # Only add legend to the first plot
        if idx == 0:
            ax.legend(fontsize=9, loc='lower right')

    # Remove any empty subplots
    for idx in range(num_plots, len(axes)):
        fig.delaxes(axes[idx])

    plt.tight_layout()
    plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
import seaborn as sns
import numpy as np

def plot_loocv_metric(df, metric_col, title, metric_type='R2', custom_color=False):
    """
    Generates a publication-ready horizontal bar plot for Leave-One-Out validation metrics.
    Supports switching between 'R2' and 'MAE', and custom theme colors.
    """
    sns.set_theme(style="white", context="talk")
    
    # Sort dataframe based on metric type so "best" value is at the top
    if metric_type == 'R2':
        df_sorted = df.sort_values(metric_col, ascending=True).copy()
    else:  # MAE (lower is better, so put lowest at the top)
        df_sorted = df.sort_values(metric_col, ascending=False).copy()
    
    # Extract names and values
    names = df_sorted['main_name'].values
    vals = df_sorted[metric_col].values
    
    # Calculate mean for the dashed line
    mean_val = np.mean(vals)
    std_val = np.std(vals)
    
    fig, ax = plt.subplots(figsize=(14, 10))
    
    # Handle color mapping
    if metric_type == 'R2':
        plot_vals = np.maximum(vals, 0)
        norm = plt.Normalize(0, 1)
        if custom_color:
            # Custom palette: Bad to Good
            colors_list = ['#843b0a', '#2a3644', '#1f3863']
            cmap = mcolors.LinearSegmentedColormap.from_list("custom_cmap", colors_list)
        else:
            cmap = plt.get_cmap('viridis')
    else:
        plot_vals = vals
        norm = plt.Normalize(0, np.max(vals))
        if custom_color:
            # Custom palette for MAE: Good (lower) to Bad (higher), reversed interpretation
            colors_list = ['#1f3863', '#2a3644', '#843b0a']
            cmap = mcolors.LinearSegmentedColormap.from_list("custom_cmap", colors_list)
        else:
            cmap = plt.get_cmap('viridis_r')
            
    colors = [cmap(norm(val)) for val in plot_vals]
    
    # Plotting the bars
    bars = ax.barh(np.arange(len(names)), plot_vals, color=colors, edgecolor='black', linewidth=1.2, height=0.8)
    
    # Add textual labels (<0 or actual rounded values)
    for idx, (bar, val) in enumerate(zip(bars, vals)):
        width = bar.get_width()
        if metric_type == 'R2' and val < 0:
            label_text = '<0'
            ax.text(0.01, bar.get_y() + bar.get_height()/2, label_text, va='center', ha='left', fontsize=11, color='black', weight='medium')
        else:
            label_text = f'{val:.3f}'
            offset = 0.01
            if metric_type == 'MAE':
                offset = np.max(vals) * 0.01
            ax.text(width + offset, bar.get_y() + bar.get_height()/2, label_text, va='center', ha='left', fontsize=11, color='#333333')

    # Add vertical mean dashed line
    ax.axvline(mean_val, color='black', linestyle='--', linewidth=2.5, label=f'Mean shown {metric_type} = {mean_val:.3f}')
    
    # Axis formatting
    ax.set_title(title, fontsize=20, fontweight='bold', pad=20)
    
    if metric_type == 'R2':
        ax.set_xlabel('Test R2 (negative values displayed as <0)', fontsize=16)
        ax.set_xlim(0, 1.05)
    else:
        ax.set_xlabel('Test Mean Absolute Error', fontsize=16)
        ax.set_xlim(0, np.max(vals) * 1.1)
        
    ax.set_ylabel('Dropped Compositions', fontsize=16)
    
    # Customizing spines and ticks (Clean, professional look)
    sns.despine(left=False, bottom=False, top=True, right=True)
    ax.spines['left'].set_color('#bbbbbb')
    ax.spines['bottom'].set_color('#bbbbbb')
    ax.spines['left'].set_linewidth(2)
    ax.spines['bottom'].set_linewidth(2)
    
    # Add light grid strictly for x-axis to track performance
    ax.xaxis.grid(True, linestyle='--', alpha=0.5, color='#dddddd')
    ax.set_axisbelow(True) # Put grid behind bars
    
    ax.tick_params(axis='x', which='major', labelsize=14, pad=10)
    ax.tick_params(axis='y', length=0) # hide ticks on y axis, just keep labels
    
    ax.set_yticks(np.arange(len(names)))
    ax.set_yticklabels(names, fontsize=13)
    
    # Styling Legend specifically
    legend = ax.legend(loc='lower right', bbox_to_anchor=(0.98, 0.60), fontsize=16, frameon=True, shadow=False, facecolor='white', borderpad=1.0)
    legend.get_frame().set_edgecolor('#cccccc')
    legend.get_frame().set_linewidth(1.5)
    
    # Add optional stats text box directly under the legend for MAE
    if metric_type == 'MAE':
        r2_col = metric_col.replace('MAE', 'R2')
        mean_r2 = df[r2_col].mean() if r2_col in df.columns else np.nan
        
        stats_text = f"Mean R² = {mean_r2:.3f}\nStd MAE = {std_val:.3f}"
        
        ax.text(0.9, 0.55, stats_text, transform=ax.transAxes, 
                fontsize=14, va='top', ha='right', weight='medium',
                bbox=dict(boxstyle='round,pad=0.6', facecolor='white', edgecolor='#cccccc', linewidth=1.5))
    
    plt.tight_layout()
    plt.show()

# -------------------------------------------------------------
# Choose metric: 'R2' or 'MAE'
# -------------------------------------------------------------
CHOSEN_METRIC = 'MAE'  # <--- Change this to 'R2' to plot R2 instead
CUSTOM_COLORS = True   # <--- Change this to False to use the default Viridis

# Generate Publication-Ready Plot for SVR and GPR Hybrid Models
if CHOSEN_METRIC == 'R2':
    plot_loocv_metric(results_df, 'Hybrid_SVR_Test_R2', 'Leave-One-Out Performance (SVR-Gompertz)', metric_type='R2', custom_color=CUSTOM_COLORS)
    plot_loocv_metric(results_df, 'Hybrid_GPR_Test_R2', 'Leave-One-Out Performance (GPR-Gompertz)', metric_type='R2', custom_color=CUSTOM_COLORS)
else:
    plot_loocv_metric(results_df, 'Hybrid_SVR_Test_MAE', 'Leave-One-Out Performance (SVR-Gompertz)', metric_type='MAE', custom_color=CUSTOM_COLORS)
    plot_loocv_metric(results_df, 'Hybrid_GPR_Test_MAE', 'Leave-One-Out Performance (GPR-Gompertz)', metric_type='MAE', custom_color=CUSTOM_COLORS)
```

# Section 0: using optuna to find the best hyperparameters for the model

```python
import optuna
from sklearn.model_selection import KFold
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import RBF, ConstantKernel as C, WhiteKernel
from sklearn.multioutput import MultiOutputRegressor
from sklearn.metrics import mean_absolute_error
from sklearn.preprocessing import StandardScaler
import numpy as np
import warnings

warnings.filterwarnings('ignore') # Suppress convergence warnings during search

print("--- Running Optuna to find Best GPR Hyperparameters on the Base Gompertz Data ---")

# Use X and Y from the earlier base Gompertz extraction
X_opt_log = np.log1p(X)
Y_opt_log = np.log1p(Y)

sc_X_opt = StandardScaler()
sc_Y_opt = StandardScaler()

X_opt_scaled = sc_X_opt.fit_transform(X_opt_log)
Y_opt_scaled = sc_Y_opt.fit_transform(Y_opt_log)

def objective(trial):
    # Propose hyperparameters
    c_val = trial.suggest_float('c_val', 1e-2, 10.0, log=True)
    length_scale = trial.suggest_float('length_scale', 1e-1, 100.0, log=True)
    noise_level = trial.suggest_float('noise_level', 1e-3, 1.0, log=True)
    n_restarts = trial.suggest_int('n_restarts_optimizer', 0, 5)
    
    # K-Fold CV to minimize MAE (using 3 or 5 splits for speed)
    kf = KFold(n_splits=3, shuffle=True, random_state=42)
    mae_fold = []
    
    for train_idx, val_idx in kf.split(X_opt_scaled):
        X_t, X_v = X_opt_scaled[train_idx], X_opt_scaled[val_idx]
        Y_t, Y_v = Y_opt_scaled[train_idx], Y_opt_scaled[val_idx]
        
        kernel = C(c_val) * RBF(length_scale) + WhiteKernel(noise_level=noise_level)
        model = MultiOutputRegressor(GaussianProcessRegressor(
            kernel=kernel, 
            n_restarts_optimizer=n_restarts, 
            random_state=42
        ))
        
        try:
            model.fit(X_t, Y_t)
            Y_pred_scaled = model.predict(X_v)
            
            # Unscale to evaluate heavily on real biological target scale
            Y_pred = np.expm1(sc_Y_opt.inverse_transform(Y_pred_scaled))
            Y_val = np.expm1(sc_Y_opt.inverse_transform(Y_v))
            mae_fold.append(mean_absolute_error(Y_val, Y_pred))
        except Exception:
            return float('inf') # Reject params that cause extreme instability
            
    return np.mean(mae_fold) if mae_fold else float('inf')

study = optuna.create_study(direction='minimize')
# 30 trials usually reaches a very solid baseline quickly for GPR bounds
study.optimize(objective, n_trials=100)

print("\n=== Optimized GPR Parameters ===")
for k, v in study.best_params.items():
    print(f"  {k}: {v}")

def get_best_gpr_model():
    """
    Returns a fresh MultiOutputRegressor(GPR) instance using Optuna's best params 
    for use dynamically throughout the rest of the notebook.
    """
    kernel = C(study.best_params['c_val'], (1e-3, 1e3)) * \
             RBF(study.best_params['length_scale'], (1e-2, 1e2)) + \
             WhiteKernel(noise_level=study.best_params['noise_level'])
             
    return MultiOutputRegressor(GaussianProcessRegressor(
        kernel=kernel, 
        n_restarts_optimizer=study.best_params['n_restarts_optimizer'], 
        random_state=42
    ))
```

# Section 2: Ensemble Extractor (Gompertz, Logistic, Transference)

```python
# ========================================================
# Phase 1: Robust Ground Truth Extraction via Ensemble
# ========================================================

def transference(t, P, R_m, lam):
    """
    Eq 2: Transference function model.
    Uses the same parameters as Gompertz: P, R_m, lam.
    """
    # Prevent division by zero if P optimizer tests 0
    P_safe = np.maximum(P, 1e-9) 
    
    exponent = np.clip(-(R_m / P_safe) * (t - lam), -700, 700)
    y = P * (1 - np.exp(exponent))
    return np.maximum(0, y)

def modified_logistic(t, P, R_m, lam):
    """
    Eq 4: Modified logistic model.
    Uses the same parameters as Gompertz: P, R_m, lam.
    """
    P_safe = np.maximum(P, 1e-9)
    
    exponent = np.clip(((4 * R_m) / P_safe) * (lam - t) + 2, -700, 700)
    return P / (1 + np.exp(exponent))

# Gompertz is already defined previously, but redeclared strictly for scope clarity
def gompertz(t, P, R_m, lam):
    e = np.exp(1)
    return P * np.exp(-np.exp((R_m * e / P) * (lam - t) + 1))


ens_features = []
ens_targets = []
ens_names = []
best_models_log = []

for name, group in data.groupby('main_name'):
    t_raw = group['Time'].values
    V_raw = group['Volume'].values
    
    def get_scalar(col):
        val = group[col].iloc[0]
        if isinstance(val, pd.Series):
            return float(val.iloc[0])
        return float(val)
        
    ts_mix = get_scalar('TS_mix_pct')
    vs_mix = get_scalar('VS_mix_pct')
    cn_mix = get_scalar('CN_ratio_mix')
    isr = get_scalar('ISR')
    
    by = get_scalar('BY')
    cd = get_scalar('CD')
    sw = get_scalar('SW_%')
    kw = get_scalar('KW_%')
    lf = get_scalar('LF_%')
    st = get_scalar('ST_%')
    h_mix = get_scalar('H_mix_pct')
    c_mix = get_scalar('C_mix_pct')
    o_mix = get_scalar('O_mix_pct')
    n_mix = get_scalar('N_mix_pct')
    
    # Intelligent Intial Guesses
    P_guess = np.max(V_raw) if np.max(V_raw) > 0 else 1.0
    dV = np.diff(V_raw)
    dt = np.diff(t_raw)
    dt[dt == 0] = 1e-6
    slopes = dV / dt
    Rm_guess = np.max(slopes) if (len(slopes) > 0 and np.max(slopes) > 0) else 1.0
    threshold = 0.05 * P_guess
    lam_indices = np.where(V_raw > threshold)[0]
    lam_guess = t_raw[lam_indices[0]] if len(lam_indices) > 0 else 0.0
    
    funcs = [
        (gompertz, "Gompertz"),
        (modified_logistic, "Modified Logistic"),
        (transference, "Transference")
    ]
    
    best_r2 = -np.inf
    best_popt = None
    best_name = None
    
    for func, fname in funcs:
        try:
            popt, _ = curve_fit(
                func, 
                t_raw, 
                V_raw, 
                p0=[P_guess, Rm_guess, lam_guess], 
                bounds=(0, [np.inf, np.inf, np.inf]), 
                maxfev=15000
            )
            pred = func(t_raw, *popt)
            r2 = r2_score(V_raw, pred)
            
            if r2 > best_r2:
                best_r2 = r2
                best_popt = popt
                best_name = fname
        except Exception as e:
            # Silently pass fails, the other models will catch it.
            pass
            
    if best_popt is not None:
        ens_features.append([ts_mix, vs_mix, cn_mix, isr, by, cd, h_mix, c_mix, o_mix, n_mix])
        ens_targets.append(best_popt)
        ens_names.append(name)
        best_models_log.append({
            'Composition': name, 
            'Winning Extraction Model': best_name, 
            'Optimized R2': best_r2, 
            'P': best_popt[0], 
            'Rm': best_popt[1], 
            'lam': best_popt[2]
        })
    else:
        print(f"CRITICAL WARNING: All three extraction models failed to converge for {name}")

X_ens = np.array(ens_features, dtype=float)
Y_ens = np.array(ens_targets, dtype=float)
unique_names_ens = np.array(ens_names)

print(f"Successfully utilized Ensemble Logic to extract PERFECT kinetic data for {len(X_ens)} reactors.")
display(pd.DataFrame(best_models_log))

# ========================================================
# Phase 1.5: LOOCV on Perfect Kinetic Data for all 3 Models
# ========================================================
print("\n--- Running LOOCV for GPR using PERFECT Kinetic Targets for 3 Models ---")
perf_results = []

for dropped_name in set(unique_names_ens):
    train_mask = unique_names_ens != dropped_name
    test_mask = unique_names_ens == dropped_name
    
    X_train, Y_train = X_ens[train_mask], Y_ens[train_mask]
    X_test, Y_test = X_ens[test_mask], Y_ens[test_mask]
    
    # Scale via Log1p
    X_train_log = np.log1p(X_train)
    Y_train_log = np.log1p(Y_train)
    X_test_log = np.log1p(X_test)
    
    sc_X, sc_Y = StandardScaler(), StandardScaler()
    X_train_scaled = sc_X.fit_transform(X_train_log)
    Y_train_scaled = sc_Y.fit_transform(Y_train_log)
    X_test_scaled = sc_X.transform(X_test_log)
    
    # Load the globally optimized GPR model from Optuna
    model_perf = get_best_gpr_model()
    model_perf.fit(X_train_scaled, Y_train_scaled)
    
    y_pred = np.expm1(sc_Y.inverse_transform(model_perf.predict(X_test_scaled)))[0]
    P_pred, Rm_pred, lam_pred = y_pred
    
    # Evaluate against the dropped test composition
    df_test = data[data['main_name'] == dropped_name].sort_values(by='Time')
    t_act = df_test['Time'].values
    V_act = df_test['Volume'].values
    
    if len(t_act) == 0: continue
        
    V_gomp = gompertz(t_act, P_pred, Rm_pred, lam_pred)
    V_logi = modified_logistic(t_act, P_pred, Rm_pred, lam_pred)
    V_tran = transference(t_act, P_pred, Rm_pred, lam_pred)
    
    perf_results.append({
        "main_name": dropped_name,
        "Perf_Gomp_R2": r2_score(V_act, V_gomp) if len(V_act) > 1 else np.nan,
        "Perf_Gomp_MAE": mean_absolute_error(V_act, V_gomp) if len(V_act) > 0 else np.nan,
        "Perf_Logi_R2": r2_score(V_act, V_logi) if len(V_act) > 1 else np.nan,
        "Perf_Logi_MAE": mean_absolute_error(V_act, V_logi) if len(V_act) > 0 else np.nan,
        "Perf_Tran_R2": r2_score(V_act, V_tran) if len(V_act) > 1 else np.nan,
        "Perf_Tran_MAE": mean_absolute_error(V_act, V_tran) if len(V_act) > 0 else np.nan,
    })

perf_df = pd.DataFrame(perf_results)

print("\nGenerating Horizontal Bar Charts for Perfect-Extraction GPR...")
try:
    if CHOSEN_METRIC == 'R2':
        plot_loocv_metric(perf_df, 'Perf_Gomp_R2', 'GPR-Gompertz model on Ensemble Kinetic Data  ', metric_type='R2', custom_color=CUSTOM_COLORS)
        plot_loocv_metric(perf_df, 'Perf_Logi_R2', 'GPR-Logistic model on Ensemble Kinetic Data   ', metric_type='R2', custom_color=CUSTOM_COLORS)
        plot_loocv_metric(perf_df, 'Perf_Tran_R2', 'GPR-Transference model on Ensemble Kinetic Data ', metric_type='R2', custom_color=CUSTOM_COLORS)
    else:
        plot_loocv_metric(perf_df, 'Perf_Gomp_MAE', 'GPR-Gompertz model on Ensemble Kinetic Data  ', metric_type='MAE', custom_color=CUSTOM_COLORS)
        plot_loocv_metric(perf_df, 'Perf_Logi_MAE', 'GPR-Logistic model on Ensemble Kinetic Data   ', metric_type='MAE', custom_color=CUSTOM_COLORS)
        plot_loocv_metric(perf_df, 'Perf_Tran_MAE', 'GPR-Transference model on Ensemble Kinetic Data ', metric_type='MAE', custom_color=CUSTOM_COLORS)
except NameError:
    print("Please ensure CHOSEN_METRIC, CUSTOM_COLORS, and plot_loocv_metric from Cell 6 are already run and loaded into memory!")
```

```python
# ==============================================================================
# Setup & Package Installation for LightGBM, TabPFN, and Optuna
# ==============================================================================

import lightgbm as lgb
from tabpfn import TabPFNRegressor
from xgboost import XGBRegressor
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import RBF, ConstantKernel as C, WhiteKernel
from sklearn.model_selection import GroupKFold
from sklearn.metrics import mean_absolute_error, r2_score
from sklearn.preprocessing import StandardScaler
import optuna
import numpy as np
import pandas as pd
import warnings

warnings.filterwarnings('ignore')

# 1. Define Direct Features: Predicting Volume strictly from Composition + Time
direct_features = ['Time', 'TS_mix_pct', 'VS_mix_pct', 'CN_ratio_mix', 'ISR', 
                   'BY', 'CD', 'H_mix_pct', 'C_mix_pct', 'O_mix_pct', 'N_mix_pct' , 'LF_%', 'ST_%',
                   'SW_%', 'KW_%']

X_dir = data[direct_features].values
y_dir = data['Volume'].values
groups_dir = data['main_name'].values

# 2. Pre-scale for GPR and TabPFN (which prefer centered data)
sc_X_dir = StandardScaler()
sc_y_dir = StandardScaler()

X_dir_sc = sc_X_dir.fit_transform(X_dir)
y_dir_sc = sc_y_dir.fit_transform(y_dir.reshape(-1, 1)).flatten()

print("Data scaled and prepared for pure Direct Time-Series Prediction.")
```

```python
# ==============================================================================
# Model 1: XGBoost (Direct Prediction with Light Optuna)
# ==============================================================================
print("--- 1. Direct Time-Series Prediction: XGBoost Tuning ---")

def objective_xgb(trial):
    params = {
        'tree_method': 'hist',
        'device': 'cuda',
        'n_estimators': trial.suggest_int('n_estimators', 50, 200),
        'max_depth': trial.suggest_int('max_depth', 3, 8),
        'learning_rate': trial.suggest_float('learning_rate', 0.01, 0.2, log=True),
        'subsample': trial.suggest_float('subsample', 0.6, 1.0),
        'random_state': 42
    }
    model = XGBRegressor(**params)
    gkf = GroupKFold(n_splits=3)
    maes = []
    
    # Train across 3 folds making sure full series are grouped intact
    for train_idx, val_idx in gkf.split(X_dir, y_dir, groups=groups_dir):
        model.fit(X_dir[train_idx], y_dir[train_idx])
        preds = model.predict(X_dir[val_idx])
        maes.append(mean_absolute_error(y_dir[val_idx], preds))
        
    return np.mean(maes)

study_xgb = optuna.create_study(direction='minimize')
study_xgb.optimize(objective_xgb, n_trials=100) # Fast tuning

print(f"\nBest XGB Params: {study_xgb.best_params}")

best_xgb = XGBRegressor(**study_xgb.best_params, random_state=42, tree_method='hist', device='cuda')
xgb_direct_results = []

print("\nRunning Full OOD LOOCV for XGBoost...")
for dropped_name in set(groups_dir):
    train_mask = groups_dir != dropped_name
    test_mask = groups_dir == dropped_name
    
    best_xgb.fit(X_dir[train_mask], y_dir[train_mask])
    preds = best_xgb.predict(X_dir[test_mask])
    
    xgb_direct_results.append({
        "main_name": dropped_name,
        "XGB_Direct_MAE": mean_absolute_error(y_dir[test_mask], preds),
        "XGB_Direct_R2": r2_score(y_dir[test_mask], preds) if sum(test_mask)>1 else np.nan
    })

xgb_dir_df = pd.DataFrame(xgb_direct_results)
display(xgb_dir_df)
```

```python
# ==============================================================================
# Model 4: TabPFN (Direct Prediction via Pre-trained Tabular Transformer)
# ==============================================================================
print("--- 4. Direct Time-Series Prediction: TabPFN ---")
print("Running OOD LOOCV for TabPFN (No Optuna Tuning Required) ...")

tabpfn_direct_results = []

for dropped_name in set(groups_dir):
    train_mask = groups_dir != dropped_name
    test_mask = groups_dir == dropped_name
    
    # Use scaled configurations, it generalizes radically better on TabPFN
    X_t, y_t = X_dir_sc[train_mask], y_dir_sc[train_mask]
    X_v, y_v = X_dir_sc[test_mask], y_dir[test_mask] # Original target for error calculate
    
    # TabPFN constraint handling (Some older CPU versions crash hard past 1k training nodes)
    # We randomly subsample to exactly 1024 safe rows to avoid overflow during multi-variate modeling 
    if len(X_t) > 1024:
        np.random.seed(42)
        idx = np.random.choice(len(X_t), 1024, replace=False)
        X_t, y_t = X_t[idx], y_t[idx]
        
    # Standard PFN configuration defaults mapped globally 
    model_tab = TabPFNRegressor(device='cuda')
    model_tab.fit(X_t, y_t)
    
    preds_sc = model_tab.predict(X_v)
    preds = sc_y_dir.inverse_transform(preds_sc.reshape(-1, 1)).flatten()
    
    tabpfn_direct_results.append({
        "main_name": dropped_name,
        "TabPFN_Direct_MAE": mean_absolute_error(y_v, preds),
        "TabPFN_Direct_R2": r2_score(y_v, preds) if len(y_v)>1 else np.nan
    })

tabpfn_dir_df = pd.DataFrame(tabpfn_direct_results)
display(tabpfn_dir_df.head(3))
```

```python
# ==============================================================================
# Model 5: PyTorch Neural Network (Mish + LayerNorm, Extensive Tuning)
# ==============================================================================
print("--- 5. Direct Time-Series Prediction: PyTorch NN Tuning ---")
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset

# Modern SOTA architecture elements: Mish activation, LayerNorm for deep stable learning
class ModernNN(nn.Module):
    def __init__(self, input_dim, hidden_dim, num_layers, dropout_rate):
        super(ModernNN, self).__init__()
        
        layers = []
        # Input layer
        layers.append(nn.Linear(input_dim, hidden_dim))
        layers.append(nn.LayerNorm(hidden_dim))
        layers.append(nn.Mish())
        layers.append(nn.Dropout(dropout_rate))
        
        # Deep residual-like hidden layers
        for _ in range(num_layers - 1):
            layers.append(nn.Linear(hidden_dim, hidden_dim))
            layers.append(nn.LayerNorm(hidden_dim))
            layers.append(nn.Mish())
            layers.append(nn.Dropout(dropout_rate))
            
        # Output scaling
        layers.append(nn.Linear(hidden_dim, 1))
        
        self.net = nn.Sequential(*layers)
        
    def forward(self, x):
        return self.net(x)

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

def objective_nn(trial):
    # Extensive SOTA tuning bounds
    hidden_dim = trial.suggest_categorical('hidden_dim', [32, 64, 128, 256])
    num_layers = trial.suggest_int('num_layers', 2, 5)
    dropout_rate = trial.suggest_float('dropout_rate', 0.05, 0.4)
    lr = trial.suggest_float('lr', 1e-4, 1e-2, log=True)
    weight_decay = trial.suggest_float('weight_decay', 1e-6, 1e-3, log=True)
    batch_size = trial.suggest_categorical('batch_size', [16, 32, 64])
    epochs = 150
    
    gkf = GroupKFold(n_splits=3)
    maes = []
    
    for train_idx, val_idx in gkf.split(X_dir_sc, y_dir_sc, groups=groups_dir):
        # Prepare tensors
        X_t = torch.FloatTensor(X_dir_sc[train_idx]).to(device)
        y_t = torch.FloatTensor(y_dir_sc[train_idx]).unsqueeze(1).to(device)
        X_v = torch.FloatTensor(X_dir_sc[val_idx]).to(device)
        
        train_loader = DataLoader(TensorDataset(X_t, y_t), batch_size=batch_size, shuffle=True)
        
        model = ModernNN(X_dir_sc.shape[1], hidden_dim, num_layers, dropout_rate).to(device)
        criterion = nn.L1Loss() # MAE Loss natively
        optimizer = optim.AdamW(model.parameters(), lr=lr, weight_decay=weight_decay)
        scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', factor=0.5, patience=15)
        
        model.train()
        for epoch in range(epochs):
            for bx, by in train_loader:
                optimizer.zero_grad()
                pred = model(bx)
                loss = criterion(pred, by)
                loss.backward()
                optimizer.step()
            
            # Step scheduler on validation pseudo-loss
            with torch.no_grad():
                model.eval()
                val_pred_sc = model(X_v)
                val_loss = criterion(val_pred_sc, torch.FloatTensor(y_dir_sc[val_idx]).unsqueeze(1).to(device))
                scheduler.step(val_loss)
                model.train()
                
        # Final predict
        model.eval()
        with torch.no_grad():
            preds_sc = model(X_v).cpu().numpy()
            
        preds = sc_y_dir.inverse_transform(preds_sc).flatten()
        y_v_real = sc_y_dir.inverse_transform(y_dir_sc[val_idx].reshape(-1, 1)).flatten()
        maes.append(mean_absolute_error(y_v_real, preds))
        
    return np.mean(maes)

study_nn = optuna.create_study(direction='minimize')
study_nn.optimize(objective_nn, n_trials=30) 

print(f"\nBest PyTorch NN Params: {study_nn.best_params}")

# Train the best NN across the OOD LOOCV loop
nn_direct_results = []
print("\nRunning Full OOD LOOCV for SOTA PyTorch NN...")

for dropped_name in set(groups_dir):
    train_mask = groups_dir != dropped_name
    test_mask = groups_dir == dropped_name
    
    # Needs scaled features
    X_t = torch.FloatTensor(X_dir_sc[train_mask]).to(device)
    y_t = torch.FloatTensor(y_dir_sc[train_mask]).unsqueeze(1).to(device)
    X_v = torch.FloatTensor(X_dir_sc[test_mask]).to(device)
    
    bs = study_nn.best_params['batch_size']
    train_loader = DataLoader(TensorDataset(X_t, y_t), batch_size=bs, shuffle=True)
    
    best_nn = ModernNN(
        X_dir_sc.shape[1], 
        study_nn.best_params['hidden_dim'], 
        study_nn.best_params['num_layers'], 
        study_nn.best_params['dropout_rate']
    ).to(device)
    
    criterion = nn.L1Loss()
    optimizer = optim.AdamW(
        best_nn.parameters(), 
        lr=study_nn.best_params['lr'], 
        weight_decay=study_nn.best_params['weight_decay']
    )
    
    best_nn.train()
    for _ in range(200): # Build deeper out-of-fold generalization
        for bx, by in train_loader:
            optimizer.zero_grad()
            loss = criterion(best_nn(bx), by)
            loss.backward()
            optimizer.step()
            
    best_nn.eval()
    with torch.no_grad():
        preds_sc = best_nn(X_v).cpu().numpy()
        
    preds = sc_y_dir.inverse_transform(preds_sc).flatten()
    
    nn_direct_results.append({
        "main_name": dropped_name,
        "NN_Direct_MAE": mean_absolute_error(y_dir[test_mask], preds),
        "NN_Direct_R2": r2_score(y_dir[test_mask], preds) if sum(test_mask)>1 else np.nan
    })

nn_dir_df = pd.DataFrame(nn_direct_results)
display(nn_dir_df.head(3))
```
