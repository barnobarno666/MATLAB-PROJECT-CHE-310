```python
import requests
import pandas as pd
from tqdm.auto import tqdm

start = "20190827"
end = "20260219"

all_data = []

for name, lat, lon in tqdm(districts, desc="Fetching districts"):
    url = (
        "https://power.larc.nasa.gov/api/temporal/daily/point"
        f"?parameters=T2M,PRECTOTCORR,RH2M,QV2M,WS2M,PS,ALLSKY_SFC_SW_DWN"
        f"&community=AG"
        f"&latitude={lat}"
        f"&longitude={lon}"
        f"&start={start}"
        f"&end={end}"
        f"&format=JSON"
    )

    r = requests.get(url, timeout=30)
    r.raise_for_status()
    data = r.json()

    df = pd.DataFrame(data["properties"]["parameter"]).reset_index()
    df = df.rename(columns={"index": "Date"})
    df["Date"] = pd.to_datetime(df["Date"], format="%Y%m%d", errors="coerce").dt.strftime("%Y-%m-%d")
    df["district"] = name

    all_data.append(df)

final = pd.concat(all_data, ignore_index=True)
final = final[["Date", "district"] + [c for c in final.columns if c not in ["Date", "district"]]]

output_file = "meteorological_data_district.csv"
final.to_csv(output_file, index=False)

print(f"Saved: {output_file}")
final.head()
```

```python
features = [
    "T2M", "PRECTOTCORR", "RH2M", "QV2M", "WS2M", "PS", "ALLSKY_SFC_SW_DWN"
]

main_tmp = main_data.copy()
final_tmp = final.copy()

main_tmp["Date"] = pd.to_datetime(main_tmp["Date"], errors="coerce").dt.strftime("%Y-%m-%d")
final_tmp["Date"] = pd.to_datetime(final_tmp["Date"], errors="coerce").dt.strftime("%Y-%m-%d")

main_tmp["_district_key"] = main_tmp["District"].astype(str).str.strip().str.lower()
final_tmp["_district_key"] = final_tmp["district"].astype(str).str.strip().str.lower()

main_tmp["_date_key"] = main_tmp["Date"]
final_tmp["_date_key"] = final_tmp["Date"]

final_features = (
    final_tmp[["_district_key", "_date_key"] + features]
    .drop_duplicates(subset=["_district_key", "_date_key"], keep="first")
)

rows_before = len(main_tmp)

main_enriched = main_tmp.merge(
    final_features,
    on=["_district_key", "_date_key"],
    how="inner"
).drop(columns=["_district_key", "_date_key"])

rows_after = len(main_enriched)
dropped_rows = rows_before - rows_after

main_data = main_enriched

output_file = "main_data_with_meteorological_features.csv"
main_data.to_csv(output_file, index=False)

print(f"Rows before merge: {rows_before}")
print(f"Rows after merge:  {rows_after}")
print(f"Dropped rows:      {dropped_rows}")
print(f"Saved: {output_file}")

main_data.head()
```

```python
import pandas as pd

# 1. Prepare the Date Column
# Ensure it is a datetime object and sorted (crucial for merge_asof)
main_data['Date'] = pd.to_datetime(main_data['Date'])
main_data = main_data.sort_values('Date').reset_index(drop=True)

# 2. Define Variables
precip_col = 'PRECTOTCORR'
mean_cols = ['T2M', 'RH2M', 'QV2M', 'WS2M', 'PS', 'ALLSKY_SFC_SW_DWN']
windows_days = [14, 28]

# 3. Calculate Time-Aware Rolling Features
# Using '14D' ensures the window covers exactly 14 calendar days, regardless of missing rows
for window in windows_days:
    window_str = f'{window}D'
    
    # Cumulative Rainfall (Sum)
    main_data[f'{precip_col}_rollsum_{window}d'] = (
        main_data.rolling(window_str, on='Date')[precip_col].sum()
    )
    
    # Averages for Temperature, Humidity, Wind, etc.
    for col in mean_cols:
        main_data[f'{col}_rollmean_{window}d'] = (
            main_data.rolling(window_str, on='Date')[col].mean()
        )

# 4. Create a Lookup Table
# We isolate the rolling features into a lookup table so we can search it for nearest dates
rolling_cols = [c for c in main_data.columns if 'rollsum' in c or 'rollmean' in c]
lookup_df = main_data[['Date'] + rolling_cols].copy()
lookup_df = lookup_df.sort_values('Date') # Must be sorted for merge_asof

# 5. Define targeted Lags mapping based on the literature
lag_mapping = {
    'PRECTOTCORR': [28, 56, 84, 112], # 4, 8, 12, 16 weeks
    'T2M': [14, 28, 56],              # 2, 4, 8 weeks
    'RH2M': [14, 28, 56],
    'QV2M': [14, 28, 56],
    'WS2M': [14, 28],                 # 2, 4 weeks
    'PS': [14, 28],
    'ALLSKY_SFC_SW_DWN': [14, 28]
}

# 6. Apply Time-Aware Lags using Nearest Matching
for col_base, lags in lag_mapping.items():
    # Find the newly created rolling columns related to this weather variable
    target_rolling_cols = [c for c in rolling_cols if c.startswith(col_base)]
    
    for lag in lags:
        # Create a temporary dataframe to calculate the exact target date we want
        temp_df = main_data[['Date']].copy()
        temp_df['Target_Date'] = temp_df['Date'] - pd.Timedelta(days=lag)
        temp_df = temp_df.sort_values('Target_Date')
        
        # MAGIC HAPPENS HERE: Find the nearest actual date in our lookup table
        matched = pd.merge_asof(
            temp_df,
            lookup_df[['Date'] + target_rolling_cols].rename(columns={'Date': 'Matched_Date'}),
            left_on='Target_Date',
            right_on='Matched_Date',
            direction='nearest' # This forces it to pick the closest date in either direction
        )
        
        # Re-sort to match main_data's original order
        matched = matched.sort_values('Date').reset_index(drop=True)
        
        # Assign the matched data back to main_data
        for roll_col in target_rolling_cols:
            main_data[f'{roll_col}_lag_{lag}d'] = matched[roll_col]

# Check your results
print(main_data[['Date', 'T2M_rollmean_14d_lag_14d']].head())
```

```python
# Add target lag features using nearest available date within each district
import numpy as np

work_df = main_data.copy()
work_df["Date"] = pd.to_datetime(work_df["Date"], errors="coerce")

lag_specs = [
    (7, "Cases_in_24_Hours_lag_1wk"),
    (14, "Cases_in_24_Hours_lag_2wk")
]

for _, new_col in lag_specs:
    work_df[new_col] = np.nan

valid_df = work_df.dropna(subset=["Date", "District", "Cases_in_24_Hours"]).copy()

for district_name, grp in valid_df.groupby("District", sort=False):
    grp = grp.sort_values("Date").copy()
    lookup = grp[["Date", "Cases_in_24_Hours"]].rename(columns={
        "Date": "Lookup_Date",
        "Cases_in_24_Hours": "Lookup_Cases"
    })

    for lag_days, new_col in lag_specs:
        query = grp[["Date"]].copy()
        query["Target_Date"] = query["Date"] - pd.Timedelta(days=lag_days)
        query["_row_id"] = grp.index

        matched = pd.merge_asof(
            query.sort_values("Target_Date"),
            lookup.sort_values("Lookup_Date"),
            left_on="Target_Date",
            right_on="Lookup_Date",
            direction="nearest"
        )

        work_df.loc[matched["_row_id"], new_col] = matched["Lookup_Cases"].values

main_data = work_df

print(main_data[["Date", "District", "Cases_in_24_Hours", "Cases_in_24_Hours_lag_1wk", "Cases_in_24_Hours_lag_2wk"]].head())
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score


target_col = "Cases_in_24_Hours"

model_df = main_data.copy()
model_df["Date"] = pd.to_datetime(model_df["Date"], errors="coerce")
model_df = model_df.dropna(subset=["Date", target_col]).sort_values("Date").reset_index(drop=True)

test_start_date = model_df["Date"].max() - pd.DateOffset(months=4)
train_df = model_df[model_df["Date"] < test_start_date].copy()
test_df = model_df[model_df["Date"] >= test_start_date].copy()

print(f"Train period: {train_df['Date'].min().date()} to {train_df['Date'].max().date()} | rows={len(train_df):,}")
print(f"Test period : {test_df['Date'].min().date()} to {test_df['Date'].max().date()} | rows={len(test_df):,}")

drop_cols = [target_col, "Date", "Source_File"]
feature_cols = [c for c in model_df.columns if c not in drop_cols]

X_train_raw = train_df[feature_cols].copy()
X_test_raw = test_df[feature_cols].copy()
y_train = train_df[target_col].astype(float).values
y_test = test_df[target_col].astype(float).values

full_X = pd.concat([X_train_raw, X_test_raw], axis=0)
full_X = pd.get_dummies(full_X, columns=full_X.select_dtypes(include=["object", "category"]).columns, drop_first=False)
full_X = full_X.replace([np.inf, -np.inf], np.nan).fillna(0)

X_train = full_X.iloc[:len(X_train_raw), :].copy()
X_test = full_X.iloc[len(X_train_raw):, :].copy()

model = XGBRegressor(
    objective="reg:squarederror",
    n_estimators=500,
    learning_rate=0.05,
    max_depth=6,
    min_child_weight=3,
    subsample=0.8,
    colsample_bytree=0.8,
    reg_alpha=0.0,
    reg_lambda=1.0,
    random_state=42,
    n_jobs=-1
)

model.fit(X_train, y_train)
pred_test = model.predict(X_test)
pred_test = np.clip(pred_test, 0, None)

mae = mean_absolute_error(y_test, pred_test)
rmse = np.sqrt(mean_squared_error(y_test, pred_test))
r2 = r2_score(y_test, pred_test)
mape = np.mean(np.abs((y_test - pred_test) / np.clip(np.abs(y_test), 1e-8, None))) * 100

print(f"MAE : {mae:.4f}")
print(f"RMSE: {rmse:.4f}")
print(f"R²  : {r2:.4f}")
print(f"MAPE: {mape:.2f}%")

results = test_df[["Date", "District", target_col]].copy()
results["Predicted"] = pred_test
results["Residual"] = results[target_col] - results["Predicted"]
results["AbsError"] = np.abs(results["Residual"])

daily_eval = results.groupby("Date", as_index=False)[[target_col, "Predicted"]].sum()
daily_eval["Residual"] = daily_eval[target_col] - daily_eval["Predicted"]
daily_eval["AbsError"] = np.abs(daily_eval["Residual"])
daily_eval["AbsError_7d_mean"] = daily_eval["AbsError"].rolling(7, min_periods=1).mean()

district_eval = (
    results.groupby("District", as_index=False)
    .agg(
        MAE=("AbsError", "mean"),
        RMSE=("Residual", lambda x: np.sqrt(np.mean(np.square(x)))),
        MeanActual=(target_col, "mean"),
        Rows=(target_col, "size")
    )
    .sort_values("MAE", ascending=False)
)

importance_df = pd.DataFrame({
    "feature": X_train.columns,
    "importance": model.feature_importances_
}).sort_values("importance", ascending=False).head(25)

fig, axes = plt.subplots(3, 2, figsize=(20, 18))

sns.scatterplot(x=y_test, y=pred_test, alpha=0.25, s=18, ax=axes[0, 0])
diag_min = min(y_test.min(), pred_test.min())
diag_max = max(y_test.max(), pred_test.max())
axes[0, 0].plot([diag_min, diag_max], [diag_min, diag_max], "r--", lw=1.5)
axes[0, 0].set_title("Actual vs Predicted (Test Rows)")
axes[0, 0].set_xlabel("Actual Cases_in_24_Hours")
axes[0, 0].set_ylabel("Predicted Cases_in_24_Hours")

sns.histplot(results["Residual"], bins=60, kde=True, ax=axes[0, 1])
axes[0, 1].axvline(0, color="red", linestyle="--", linewidth=1.2)
axes[0, 1].set_title("Residual Distribution")
axes[0, 1].set_xlabel("Residual (Actual - Predicted)")

sns.scatterplot(x=results["Predicted"], y=results["Residual"], alpha=0.25, s=18, ax=axes[1, 0])
axes[1, 0].axhline(0, color="red", linestyle="--", linewidth=1.2)
axes[1, 0].set_title("Residuals vs Predicted")
axes[1, 0].set_xlabel("Predicted")
axes[1, 0].set_ylabel("Residual")

axes[1, 1].plot(daily_eval["Date"], daily_eval[target_col], label="Actual", linewidth=2)
axes[1, 1].plot(daily_eval["Date"], daily_eval["Predicted"], label="Predicted", linewidth=2)
axes[1, 1].set_title("Daily Total Cases: Actual vs Predicted (Test Period)")
axes[1, 1].set_xlabel("Date")
axes[1, 1].set_ylabel("Daily Total Cases")
axes[1, 1].legend()

top_districts = district_eval.head(20).sort_values("MAE", ascending=True)
axes[2, 0].barh(top_districts["District"], top_districts["MAE"])
axes[2, 0].set_title("Top 20 Districts by MAE (Higher = Harder)")
axes[2, 0].set_xlabel("MAE")

axes[2, 1].barh(importance_df["feature"][::-1], importance_df["importance"][::-1])
axes[2, 1].set_title("Top 25 Feature Importances (XGBoost)")
axes[2, 1].set_xlabel("Importance")

plt.tight_layout()
plt.show()

plt.figure(figsize=(14, 5))
plt.plot(daily_eval["Date"], daily_eval["AbsError"], label="Daily Abs Error", alpha=0.6)
plt.plot(daily_eval["Date"], daily_eval["AbsError_7d_mean"], label="7-day Mean Abs Error", linewidth=2.5)
plt.title("Error Stability Over Time (Test Period)")
plt.xlabel("Date")
plt.ylabel("Absolute Error")
plt.legend()
plt.tight_layout()
plt.show()

print("\nWorst 10 districts by MAE:")
display(district_eval.head(10))

print("\nBest 10 districts by MAE:")
display(district_eval.sort_values("MAE", ascending=True).head(10))
```
